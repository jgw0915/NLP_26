Consider the function $$f(x) = \displaystyle \frac{1}{2} x^{T}Qx + q^{T}x +10$$
where $Q$ and $c$ are given by:
$$
Q = \left[
\begin{array}{cc}
    20 & 5\\
    5 & 2
\end{array}
\right]
\hbox{ and }
q = \left[
\begin{array}{c}
    -14\\
    -6
\end{array}
\right]
$$

For this problem, $\displaystyle \frac{A}{a} = 30.234$, and
so the linear convergence rate $\delta$ of
the steepest descent algorithm is bounded above by
$\delta = \left( \displaystyle \frac{A/a-1}{A/a+1} \right)^{2} = 0.8760$.

Now we established the positive definite matrix $Q$ and the vector $q$

```python
import numpy as np
```

```python
Q = np.array([[20,5],[5,2]])
print(Q)
```

```python
q=np.array([[-14],[-6]])
print(q)
```

Set $k=1$.

```python
k=1
```

Given $x^{1}=(40,-100)$.

```python
x=[0]
x.append(np.array([[40],[-100]]))
print("x_%d =" % (k))
print("\n",x[k])
print(x)
```

Compute $d^{1}=Qx^{1}$.

```python
d=[0]
d.append(-np.dot(Q,x[k])-q)
print("d_1 = \n",d[k])
print("norm(d_1)=%f"% np.linalg.norm(d[k]))
```

Compute the step length $\alpha=\frac{d^Td}{d^tQd}$ 

```python
alpha=[0]
alpha.append(np.dot(np.transpose(d[k]),d[k])/np.dot(np.transpose(d[k]),np.dot(Q,d[k])))
print("alpha_1 = \n",alpha[k])
```

Compute the new iteration $x^{k+1}=x^{k}+\alpha^{k} d^{k}$

```python
x.append(x[k]+alpha[k]*d[k])
print("x_1 = \n",x[k+1])
```

```python
MaxIteration=60
Tol=1e-4
k=1
x=[0]
d=[0]
alpha=[0]
x.append(np.array([[40],[-100]]))
print("x_%d =" % (k))
print("\n",x[k])
LeaveLoop=0
while LeaveLoop==0:
    d.append(-np.dot(Q,x[k])-q)
    alpha.append(np.dot(np.transpose(d[k]),d[k])/np.dot(np.transpose(d[k]),np.dot(Q,d[k])))
    x.append(x[k]+alpha[k]*d[k])
    print("x_%d =" % (k))
    print("\n",x[k])
    print("|d_%d|_2 =%f\n" % (k,np.linalg.norm(d[k])))
    print("alpha_%d =" % (k))
    print("\n",alpha[k])
    if k>=MaxIteration or np.linalg.norm(d[k])<Tol:
        LeaveLoop=1
    k=k+1
```
